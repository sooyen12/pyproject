```python
import pandas as pd
import folium

# 엑셀 파일 경로 지정
file_path = '/content/drive/MyDrive/자전거1.xlsx'

# 엑셀 파일 읽기
df_bike = pd.read_excel(file_path)
```

```python
# 위도, 경도 컬럼 이름
latitude_column = 'WGS84위도'
longitude_column = 'WGS84경도'

# 방면 컬럼 정리 (공백 제거, 문자열 처리)
df_cctv['촬영방면정보'] = df_cctv['촬영방면정보'].astype(str).str.strip()

# 지도 중심을 첫 번째 좌표로 설정
center_lat = df_cctv[latitude_column].iloc[0]
center_lon = df_cctv[longitude_column].iloc[0]
```

```python
# 공백 제거 후 강남구만 뽑기
df_bike['시도시군구명'] = df_bike['시도시군구명'].astype(str).str.strip()
gangnam_df = df_bike[df_bike['시도시군구명'].str.contains('서울특별시 강남구', na=False)]
```

```python
# 자전거 사고 다발지역 좌표 컬럼명
latitude = '위도'
longitude = '경도'
```

```python
import folium
from folium import FeatureGroup, LayerControl
from IPython.display import display
display(m)

# 지도 생성
map_center = [37.5145, 127.1065]
m = folium.Map(location=map_center, zoom_start=13)

# 자전거 사고 레이어
bike_layer = FeatureGroup(name='자전거 사고 다발지역')

for _, row in gangnam_df.iterrows():
    lat = row['위도']
    lon = row['경도']
    folium.CircleMarker(
        location=[lat, lon],
        radius=5,
        color='blue',
        fill=True,
        fill_color='blue',
        fill_opacity=0.5
    ).add_to(bike_layer)

# CCTV 레이어
cctv_layer = FeatureGroup(name='CCTV 위치')

for _, row in df_cctv.iterrows():
    lat = row['WGS84위도']
    lon = row['WGS84경도']
    direction = row['촬영방면정보']

    color = 'red' if '360' in direction else 'gray'

    folium.CircleMarker(
        location=[lat, lon],
        radius=5,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.5,
        popup=direction
    ).add_to(cctv_layer)

# 레이어 추가
m.add_child(bike_layer)
m.add_child(cctv_layer)

# 레이어 컨트롤 추가
m.add_child(LayerControl())

# 저장 또는 출력
m.save('map.html')
```
